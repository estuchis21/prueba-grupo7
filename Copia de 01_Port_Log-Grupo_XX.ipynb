## Introducción y Contexto del problema

### Sprint 1

El Puerto Fluvial de Rosario es uno de los complejos portuarios más importantes de América del Sur, siendo el principal punto de exportación de granos y derivados de la Argentina. Diariamente ingresan y egresan decenas de buques de distintas banderas con cargas de diverso tipo.

Ese sistema acumuló durante décadas inconsistencias de formato en fechas, matrículas de buques y valores numéricos fuera de rango, generando registros que no pueden incorporarse directamente al nuevo sistema.

````python
#@title Conexión y Push definitivo en un solo bloque
import os

# 1. Volvemos a armar la URL con tu token secreto verificado
url_remota = f"https://{GITHUB_TOKEN}@github.com/estuchis21/prueba-grupo7.git"

print("🔗 Vinculando repositorio remoto...")
os.system('git remote remove origin > /dev/null 2>&1')
os.system(f'git remote add origin {url_remota}')

# 2. Verificamos que el remote esté activo
print("\n--- Estado actual de remotes ---")
os.system('git remote -v')

# 3. Hacemos el push directamente a la rama Sprint_1
print("\n🚀 Subiendo a GitHub...")
resultado = os.system('git push -u origin Sprint_1')

if resultado == 0:
    print("\n✅ ¡ÉXITO TOTAL! Tu código ya se encuentra en GitHub en la rama Sprint_1.")
else:
    print("\n❌ El push falló. Asegúrate de haber creado un repositorio vacío llamado 'prueba-grupo7' en tu cuenta de GitHub.")
````

````python
#@title Sincronización avanzada: Integrar main y establecer Sprint_1 como principal
import os

url_remota = f"https://{GITHUB_TOKEN}@github.com/estuchis21/prueba-grupo7.git"

print("🔄 Conectando y actualizando referencias...")
os.system(f'git remote set-url origin {url_remota}')
os.system('git fetch origin')

print("🔀 Fusionando el historial remoto (main) con nuestra rama local...")
# Esto unifica el README inicial de GitHub con tus carpetas nuevas de forma segura
os.system('git pull origin main --allow-unrelated-histories --no-edit')

print("🌿 Asegurando rama local Sprint_1...")
os.system('git branch -M Sprint_1')

print("🚀 Subiendo y reemplazando/actualizando todo en GitHub...")
resultado = os.system('git push -u origin Sprint_1 --force')

if resultado == 0:
    print("\n✅ ¡ÉXITO ROTUNDO! La rama Sprint_1 ha sido sincronizada y subida correctamente.")
    print("👉 Ve a GitHub, actualiza la página y selecciona la rama 'Sprint_1' para ver tu estructura intacta junto al README.")
else:
    print("\n❌ Hubo un inconveniente con el push. Verifica que el GITHUB_TOKEN esté activo en los secretos.")
````
